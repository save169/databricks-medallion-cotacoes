```python
#Transforma a tabela silver em dados limpos para visualização
from pyspark.sql.functions import col

CATALOGO = "workspace"
SCHEMA = "exp_dados"

df_silver = spark.read.table(f"{CATALOGO}.{SCHEMA}.silver_cotacoes_limpo")

df_gold = df_silver.select(
    col("moeda_origem"),
    col("data_cotacao"),
    col("moeda_destino"),
    col("valor_cotacao"),
    col("_data_ingestao")
)

tabela_gold = f"{CATALOGO}.{SCHEMA}.gold_cotacoes_diarias"
df_gold.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable(tabela_gold)
print("Gold concluída!")
```
